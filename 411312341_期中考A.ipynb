````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-120b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
import sqlite3

def init_db():
    conn = sqlite3.connect('events.db')
    cursor = conn.cursor()

    cursor.execute("DROP TABLE IF EXISTS Registrations")
    cursor.execute("DROP TABLE IF EXISTS Events")

    cursor.execute("""
    CREATE TABLE Events (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT UNIQUE,
        event_date TEXT,
        capacity INTEGER,
        remain_slots INTEGER
    )
    """)

    cursor.execute("""
    CREATE TABLE Registrations (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        user_name TEXT,
        event_name TEXT,
        num_people INTEGER,
        status TEXT
    )
    """)

    events_data = [
        ('迎新晚會', '2026-11-20', 6, 6),
        ('程式工作坊', '2026-11-25', 4, 4),
        ('籃球友誼賽', '2026-11-28', 3, 3)
    ]
    cursor.executemany("INSERT INTO Events (name, event_date, capacity, remain_slots) VALUES (?, ?, ?, ?)", events_data)

    conn.commit()
    conn.close()

init_db()
print("✅ 資料庫與初始資料建立完成")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def show_tables():
    conn = sqlite3.connect('events.db')
    print("【Events 活動資料表】")
    display(pd.read_sql_query("SELECT * FROM Events", conn))
    print("\n【Registrations 報名紀錄表】")
    display(pd.read_sql_query("SELECT * FROM Registrations", conn))
    conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def get_ai_intent(user_input):
    prompt = f"""請從以下句子抽出動作(報名/取消/查詢/統計)、活動(迎新晚會/程式工作坊/籃球友誼賽)與人數(數字)。
    直接輸出「動作,活動,人數」，必須用半形逗號隔開，不要有任何其他文字。若無活動請填寫「無」。
    句子：「{user_input}」"""

    try:
        response = client.chat.completions.create(
            model=MODEL,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.1
        )
        raw = response.choices[0].message.content
        if not raw:
            return "未知", "無", 0, "API回傳為空"

        # 清理字串
        raw = raw.strip().replace('，', ',').replace('「', '').replace('」', '')
        parts = raw.split(',')

        action = parts[0].strip() if len(parts) > 0 else "未知"
        event = parts[1].strip() if len(parts) > 1 else "無"
        if not event: # 處理 AI 回傳 "統計,," 的狀況
            event = "無"

        # 確保人數是數字
        num_str = "".join(filter(str.isdigit, parts[2])) if len(parts) > 2 else ""
        num_people = int(num_str) if num_str else (1 if action in ['報名', '取消'] else 0)

        # 【最終修正邏輯】
        # 1. 如果動作是統計，活動強制為「無」，人數為 0
        if action == "統計":
            event = "無"
            num_people = 0
        # 2. 如果不是統計，且活動不是指定的三個（例如歌唱大賽），則整句視為無法判斷的未知
        elif event not in ["迎新晚會", "程式工作坊", "籃球友誼賽"]:
            action = "未知"
            event = "無"
            num_people = 0

        # 3. 確保動作一定是這四個之一
        if action not in ["報名", "取消", "查詢", "統計"]:
            action = "未知"
            event = "無"
            num_people = 0

        return action, event, num_people, raw

    except Exception as e:
        return "未知", "無", 0, f"API錯誤: {str(e)}"
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
ai_tests = [
    ("我要報名迎新晚會2人", "報名", "迎新晚會", 2),
    ("我和另外兩個朋友要參加程式工作坊", "報名", "程式工作坊", 3),
    ("幫我取消籃球友誼賽", "取消", "籃球友誼賽", 1),
    ("籃球友誼賽還有幾個名額", "查詢", "籃球友誼賽", 0),
    ("請列出各活動報名狀況", "統計", "無", 0),
    ("我想報名迎新晚會4人", "報名", "迎新晚會", 4),
    ("今天天氣真不錯", "未知", "無", 0),
    ("我想要報名歌唱大賽", "未知", "無", 0)
]

passed = 0
for text, exp_act, exp_evt, exp_num in ai_tests:
    act, evt, num, raw = get_ai_intent(text)
    success = (act == exp_act and evt == exp_evt and num == exp_num)
    if success:
        passed += 1
    # 修正：移除多餘的 print()
    print(f"輸入: {text}\n  預期: {exp_act},{exp_evt},{exp_num} | 實際: {act},{evt},{num} | 原始回覆: {raw} | {'✅' if success else '❌'}\n")

print(f"總計通過: {passed} / {len(ai_tests)}")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def process_request(user_name, action, event_name, num_people):
    if action == "未知" or event_name == "無" and action not in ["統計"]:
        return "無法判斷或無此活動。目前活動有：迎新晚會、程式工作坊、籃球友誼賽。"

    conn = sqlite3.connect('events.db')
    try:
        cursor = conn.cursor()

        if action == "統計":
            cursor.execute("""
                SELECT e.name, e.event_date, e.capacity, e.remain_slots,
                       COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.num_people ELSE 0 END), 0) as total_enrolled
                FROM Events e
                LEFT JOIN Registrations r ON e.name = r.event_name
                GROUP BY e.id
            """)
            rows = cursor.fetchall()
            result = "【活動報名統計】\n"
            for row in rows:
                result += f"- {row[0]} ({row[1]}): 總名額 {row[2]}, 已報名 {row[4]}, 剩餘 {row[3]}\n"
            return result.strip()

        cursor.execute("SELECT id, capacity, remain_slots FROM Events WHERE name=?", (event_name,))
        event_record = cursor.fetchone()

        if not event_record:
            return f"找不到活動「{event_name}」。"

        event_id, capacity, remain_slots = event_record

        if action == "查詢":
            return f"「{event_name}」目前還剩下 {remain_slots} 個名額。"

        if not user_name:
            return "報名或取消需要提供姓名。"

        if action == "報名":
            if not (1 <= num_people <= 3):
                return "報名失敗！單次報名人數必須在 1～3 人之間。"

            cursor.execute("SELECT COUNT(DISTINCT event_name) FROM Registrations WHERE user_name=? AND status='已報名'", (user_name,))
            enrolled_events_count = cursor.fetchone()[0]

            cursor.execute("SELECT COUNT(*) FROM Registrations WHERE user_name=? AND event_name=? AND status='已報名'", (user_name, event_name))
            already_enrolled = cursor.fetchone()[0]

            if already_enrolled > 0:
                return f"報名失敗！您已經報名過「{event_name}」了。"

            if enrolled_events_count >= 2:
                return "報名失敗！每人最多只能報名 2 個不同的活動。"

            if remain_slots < num_people:
                return f"報名失敗！「{event_name}」名額不足，僅剩 {remain_slots} 個名額。"

            cursor.execute("UPDATE Events SET remain_slots = remain_slots - ? WHERE name=?", (num_people, event_name))
            cursor.execute("INSERT INTO Registrations (user_name, event_name, num_people, status) VALUES (?, ?, ?, '已報名')", (user_name, event_name, num_people))
            conn.commit()
            return f"報名成功！「{event_name}」剩餘 {remain_slots - num_people} 個名額。"

        if action == "取消":
            cursor.execute("SELECT id, num_people FROM Registrations WHERE user_name=? AND event_name=? AND status='已報名'", (user_name, event_name))
            reg_record = cursor.fetchone()

            if not reg_record:
                return f"取消失敗！查無您在「{event_name}」的報名紀錄。"

            reg_id, enrolled_num = reg_record

            cursor.execute("UPDATE Registrations SET status='已取消' WHERE id=?", (reg_id,))
            cursor.execute("UPDATE Events SET remain_slots = remain_slots + ? WHERE name=?", (enrolled_num, event_name))
            conn.commit()
            return f"取消成功！已釋出 {enrolled_num} 個名額。"

    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
def main_flow(user_name, user_input):
    action, event_name, num_people, _ = get_ai_intent(user_input)
    return process_request(user_name, action, event_name, num_people)

flow_tests = [
    ("陳小明", "我要報名迎新晚會2人"),
    ("陳小明", "幫我報名程式工作坊1人"),
    ("陳小明", "幫我報名籃球友誼賽"),
    ("陳小明", "幫我取消迎新晚會"),
    ("林小美", "我和3個室友報名迎新晚會"),
    ("林小美", "報名迎新晚會3人"),
    ("王大寶", "籃球友誼賽還有幾個名額"),
    ("幹部", "各活動報名狀況")
]

# 測試前確保資料狀態乾淨
init_db()

print("【流程測試開始】")
for i, (name, txt) in enumerate(flow_tests, 1):
    print(f"步驟 {i} | {name}: {txt}")
    res = main_flow(name, txt)
    print(f"系統回覆: {res}\n")

show_tables()
````
